```python
#@title  🛠️ PASO 1: Conectar Base de Datos (Dale Play) { display-mode: "form" }
import pandas as pd
import sqlite3
import time
from IPython.core.magic import register_cell_magic
from IPython.display import display, HTML, clear_output
from tqdm.notebook import tqdm
# 1. Preparación de datos
# Enlace directo al CSV en tu GitHub
url_diagnoses = "https://raw.githubusercontent.com/hectorhormazabal-tt/TT-SQL-DA-95/refs/heads/master/diagnoses_dim.csv"
url_appointments = "https://raw.githubusercontent.com/hectorhormazabal-tt/TT-SQL-DA-95/refs/heads/master/emergency_appointments.csv"
url_patients = "https://raw.githubusercontent.com/hectorhormazabal-tt/TT-SQL-DA-95/refs/heads/master/patients_dim.csv"

df_diagnoses = pd.read_csv(url_diagnoses)
df_appointments = pd.read_csv(url_appointments)
df_patients = pd.read_csv(url_patients)

connector = sqlite3.connect(':memory:')

df_diagnoses.to_sql('tt_da_95_data_hospital_diagnoses', connector, index=False, if_exists='replace')
df_appointments.to_sql('tt_da_95_data_hospital_appointments', connector, index=False, if_exists='replace')
df_patients.to_sql('tt_da_95_data_hospital_patients', connector, index=False, if_exists='replace')

# 2. Registro del comando mágico con "Show" de carga
@register_cell_magic
def sql(line, cell):
    # Mensaje de inicio con estilo
    print("🚀 Héctor's SQL Engine | Ejecutando consulta...")

    # Barra de progreso estética
    # (Simulamos un pequeño delay para que los alumnos aprecien el proceso)
    for _ in tqdm(range(10), desc="Procesando datos", leave=False, bar_format='{l_bar}{bar}| {n_fmt}/{total_fmt}'):
        time.sleep(0.1)

    try:
        resultado = pd.read_sql(cell, connector)

        # Limpiamos el mensaje de "Ejecutando" para mostrar el resultado final
        clear_output(wait=True)
        display(HTML("<b style='color: #4CAF50;'>✅ Query completed successfully:</b>"))
        return display(resultado)

    except Exception as e:
        clear_output(wait=True)
        display(HTML(f"<b style='color: #F44336;'>❌ Query execution failed:</b><br><code style='color: grey;'>{str(e)}</code>"))

print("✨ ¡Héctor, tu motor SQL ahora tiene barra de progreso y mensajes de colores!")
```
