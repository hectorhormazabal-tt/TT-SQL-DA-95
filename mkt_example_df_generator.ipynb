### __Generación de dataframe con data de marketing__
---

Script que muestra una función que permite generar un pandas dataframe de tamaño variable con datos de marketing generados artificialmente para utilizar como ejemplo y aplicar técnicas de análisis de datos.

##### __1. Definición de imports principales__
Librerías requeridas para la ejecución del script. Adicionalmente se presenta la versión de Python utilizada

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import random
import uuid
from datetime import datetime, timedelta
import string
import sys
import time
print(sys.version)
```

##### __2. Sumario del dataframe__
La siguiente función crea la base da datos de tamaño a definir. Contiene las siguientes variables para cada cliente:
* __ID__: UUID generado de forma aleatoria y única para cada cliente
* __Genero__: Género con el que se identificó el usuario al momento de registrarse (M: Masculino, F: Femenino, NB: No-Binario, SR: Sin respuesta, es decir, el cliente seleccionó no especificar su género).
* __Edad__: Medida en años
* __Continente__: Distribuidos en base a la densidad esperada de clientes sobre los continentes considerando este ejemplo (retail LATAM)
* __País__: Distribuidos en base a la densidad esperada de clientes sobre los continentes considerando este ejemplo (retail LATAM)
* __Fecha de registro del cliente__: Fecha en el que el cliente se registró en el marketplace del retail por primera vez.
* __Tipo de subscripción__: Free, Premium, Ultra. Dependiendo de esta subscripción, el cliente accede a: promociones, recomendaciones, menos anuncios, preferencias y feed personalizado, facilidades de pago, acumulación de puntos de cliente para canjear, etc.
* __Puntos clientes__: Puntaje que ganan los clientes en base a sus compras y el valor de estas.
* __Medio de pago guardado__: Medio de pago registrado (de existir) más utilizado por el cliente. También es válido no tener medio de pago registrado para compras.

##### __3. Definición de la función__

```python
def mkt_data_generator(num_clientes, semilla=None): 
    #Manejo de semilla del generador de números aleatorios.
    print(" ")
    print("DATA FRAME - MARKETING DATA GENERATOR SUMMARY: ")
    print(" ")
    if semilla is not None:
        # Fijar las semillas para la reproducibilidad
        np.random.seed(semilla)
        random.seed(semilla)
        print(f" - Semilla fijada a: {semilla}. El dataset es reproducible.")
    else:
        # Si la semilla es None, usa el tiempo actual (aleatoriedad total)
        np.random.seed()
        random.seed()
        print(" - Dataset generado con aleatoriedad total.")
    
    # GENERACIÓN DE VARIABLES DEL CLIENTE

    start = time.time()
    #ID CLIENTES
    ids_clientes = []
    for _ in range(num_clientes):
        nuevo_uuid = str(uuid.uuid4())
        ids_clientes.append(nuevo_uuid)

    #GENERO
    lista_genero     = []
    genero_set       = ['M', 'F', 'NB', 'SR'] # Masculino, Femenino, No Binario, Sin Responder
    pesos_genero = [45, 47, 5, 3] # Distribución porcentual aproximada
    for _ in range(num_clientes):
        genero = random.choices(genero_set, weights= pesos_genero, k=1)[0]
        lista_genero.append(genero)

    #EDAD
    lista_edad = []
    EDAD_MEDIA_COMPRADOR = 35   # La mayoría de los clientes se agrupan alrededor de esta edad
    DESVIACION_ESTANDAR = 10    # Controla qué tan dispersas están las edades alrededor de la media
    EDAD_MINIMA = 18            # Asegura que todos son mayores de edad
    EDAD_MAXIMA = 75            # Límite superior para reducir la cantidad de clientes muy mayores (que compran menos online)
    for i in range(num_clientes):
        while True:
            edad_simulada = random.gauss(EDAD_MEDIA_COMPRADOR, DESVIACION_ESTANDAR)
            edad = int(round(edad_simulada))
            if EDAD_MINIMA <= edad <= EDAD_MAXIMA:
                lista_edad.append(edad)
                break
    
    
    #CONTINENTE - PAIS
    datos_geograficos = {
    "América": [
        "Chile", "México", "Argentina", "Colombia", "Perú", "EE. UU.",
        "Canadá", "Brasil", "Ecuador", "Venezuela"
    ],
    "Europa": [
        "España", "Alemania", "Francia", "Italia", "Reino Unido",
        "Portugal", "Países Bajos", "Suiza", "Suecia"
    ],
    "Asia": [
        "China", "India", "Japón", "Corea del Sur", "Tailandia",
        "Vietnam", "Malasia", "Indonesia"
    ],
    "África": [
        "Sudáfrica", "Egipto", "Nigeria", "Kenia", "Marruecos"
    ],
    "Oceanía": [
        "Australia", "Nueva Zelanda"
    ]}

    lista_continentes = list(datos_geograficos.keys())
    lista_continentes = list(datos_geograficos.keys())
    pesos_continentes = [80, 15, 2, 2, 1]
    lista_continentes_output = []
    lista_paises_output = []
    for _ in range(num_clientes):
        continente_elegido = random.choices(lista_continentes, weights=pesos_continentes, k=1)[0]
        lista_continentes_output.append(continente_elegido)
        paises_validos = datos_geograficos[continente_elegido]
        pais_elegido = random.choice(paises_validos)
        lista_paises_output.append(pais_elegido)


    #FECHA REGISTRO CLIENTE
    fecha_fin = datetime(2025, 11, 2)
    fecha_inicio = fecha_fin - timedelta(days=5 * 365) 
    meses      = list(range(1, 13))
    peso_meses = [5,  3,  10, 3,  3,  3,  3,  15, 15, 15, 5, 20]
    lista_fechas_output = []
    for _ in range(num_clientes):
        mes_elegido = random.choices(meses, weights=peso_meses, k=1)[0]
        while True:
            año_elegido = random.randint(fecha_inicio.year, fecha_fin.year)
            dia_elegido = random.randint(1, 28)
            try:
                fecha_propuesta = datetime(año_elegido, mes_elegido, dia_elegido)
                    
                if fecha_inicio <= fecha_propuesta <= fecha_fin:
                    lista_fechas_output.append(fecha_propuesta.strftime("%Y-%m-%d"))
                    break
                    
            except ValueError:
                continue


    # TIPO DE SUSCRIPCIÓN
    suscripcion_opciones = ["FREE", "PREMIUM", "ULTRA"]
    suscripcion_pesos = [50, 35, 15]
    suscripciones = []
    for i in range(num_clientes):
        tipo_suscripcion = random.choices(suscripcion_opciones, weights=suscripcion_pesos, k=1)[0]
        suscripciones.append(tipo_suscripcion)


    # PUNTOS CLIENTE
    num_clientes = len(suscripciones) 
    multiplicadores_puntos = {
        "FREE": 1.0,  
        "PREMIUM": 1.5, 
        "ULTRA": 2.5    
    }
    rango_base_min = 500
    rango_base_max = 8000
    puntos_maximos = 20000
    puntos_cliente = []
    for suscripcion in suscripciones:
        puntos_base = random.randint(rango_base_min, rango_base_max)
        multiplicador = multiplicadores_puntos[suscripcion]
        puntos_finales = int(round(puntos_base * multiplicador))
        
        if puntos_finales > puntos_maximos:
            puntos_finales = puntos_maximos
        puntos_cliente.append(puntos_finales)

    #MEDIO DE PAGO GUARDADO MÁS USADO
    medios_opciones = ["Tarjeta Debido Bancario", "Tarjeta Credito Bancario" ,"Tarjeta credito marketplace", "Paypal", "Mercado Pago" ,"No agregado"]
    medios_pesos = [15, 35, 25, 10, 10, 5]
    medio_pago_guardado = []
    for i in range(num_clientes):
        medio_elegido = random.choices(medios_opciones, weights=medios_pesos, k=1)[0]
        medio_pago_guardado.append(medio_elegido)

    #NIVEL EDUCATIVO 
    educacion_opciones = ["escolar_completa", "superior_incompleta", "superior_completa", "postgrado", "escolar_incompleta"]
    educacion_pesos = [35, 25, 25, 10, 5]
    nivel_educativo = []
    for i in range(num_clientes):
        nivel_elegido = random.choices(educacion_opciones, weights=educacion_pesos, k=1)[0]
        nivel_educativo.append(nivel_elegido)


    #CUENTA CON LA TARJETA DEL RETAIL
    probabilidad_extra_si_tiene_otro = 0.15
    tiene_tarjeta_retail = []
    for medio in medio_pago_guardado:
        if medio == 'Tarjeta credito marketplace':
            tiene_tarjeta_retail.append(1)
        else:
            if random.random() < probabilidad_extra_si_tiene_otro:
                tiene_tarjeta_retail.append(1)
            else:
                tiene_tarjeta_retail.append(0)

    # TIENE DESCARGADA DE LA APP DEL RETAIL EN SU CELULAR.
    probabilidades_app = {
    "FREE": 0.35,
    "PREMIUM": 0.55,
    "ULTRA": 0.90}

    tiene_app_descargada = []
    for suscripcion in suscripciones:
        prob_descarga = probabilidades_app[suscripcion]
        if random.random() < prob_descarga:
            tiene_app_descargada.append(1)
        else:
            tiene_app_descargada.append(0)
    
    #FECHA DE INSTALACIÓN DE LA APP (DE SER APLICABLE)
    fecha_fin = datetime.now()
    fechas_registro_dt = [datetime.strptime(d, '%Y-%m-%d') for d in lista_fechas_output] 
    fecha_instalacion_app = []
    min_dias_despues_registro = 7   
    max_dias_despues_registro = 730 

    for i in range(num_clientes):
        if tiene_app_descargada[i] == 0:
            fecha_instalacion_app.append(None)
        else:
            fecha_reg = fechas_registro_dt[i]
            dias_maximos_posibles = (fecha_fin - fecha_reg).days
            
            if dias_maximos_posibles < min_dias_despues_registro:
                fecha_instalacion_app.append(fecha_reg.strftime('%Y-%m-%d'))
                continue
            retraso_valido = random.randint(min_dias_despues_registro, dias_maximos_posibles)
                
            fecha_inst = fecha_reg + timedelta(days=retraso_valido)
            fecha_instalacion_app.append(fecha_inst.strftime('%Y-%m-%d'))
    
    #NOTIFICACIONES DE LA APP ACTIVADAS 
    prob_activacion = 0.70
    notificaciones_activadas = []

    for tiene_app in tiene_app_descargada:
        if tiene_app == 0:
            notificaciones_activadas.append(0)
        else:
            if random.random() < prob_activacion:
                notificaciones_activadas.append(1)
            else:
                notificaciones_activadas.append(0)


    #SISTEMA OPERATIVO PLATAFORMA APP
    plataformas_opciones = ["iOS", "Android", "other"]
    plataformas_pesos = [45, 50, 5] 
    versiones_os = {
        "iOS": [f"iOS {random.randint(15, 18)}.{random.randint(0, 5)}" for _ in range(5)],
        "Android": [f"Android {random.randint(12, 16)}.{random.randint(0, 3)}" for _ in range(5)],
        "other": ["HarmonyOS", "Windows Phone", "KaiOS"]
    }
    plataforma_app = []

    for tiene_app in tiene_app_descargada:
        if tiene_app == 0:
            plataforma_app.append(None)
        else:
            plataforma_elegida = random.choices(plataformas_opciones, weights=plataformas_pesos, k=1)[0]
            
            if plataforma_elegida in ["iOS", "Android"]:
                
                version_os = random.choice(versiones_os[plataforma_elegida])
                plataforma_app.append(f"{plataforma_elegida} {version_os.split(' ')[1]}")
            else:
                
                plataforma_app.append(random.choice(versiones_os[plataforma_elegida]))

    # ORDENES, MONTO TOTAL, TICKET PROMEDIO 
    rango_ordenes = {
    "FREE": (8, 50),
    "PREMIUM": (5, 40),
    "ULTRA": (3, 35)
    }

    rango_ticket_promedio = {
        "FREE": (20.00, 70.00),     
        "PREMIUM": (75.00, 200.00),   
        "ULTRA": (180.00, 500.00)     
    }

    cantidad_ordenes = []
    total_gastado = []
    ticket_promedio = []

    for suscripcion in suscripciones:
        min_ord, max_ord = rango_ordenes[suscripcion]
        ordenes = random.randint(min_ord, max_ord)
        cantidad_ordenes.append(ordenes)
        
        min_ticket, max_ticket = rango_ticket_promedio[suscripcion]
        mu = (min_ticket + max_ticket) / 2
        sigma = (max_ticket - min_ticket) / 6
        ticket = round(max(min_ticket, min(max_ticket, random.gauss(mu, sigma))), 2) 
        ticket_promedio.append(ticket)
        
        total = round(ordenes * ticket, 2)
        total_gastado.append(total)

    
    # VISITAS APP Y WEB 
    rangos_visitas = {
    "NO_APP_WEB": (30, 180),
    "CON_APP_APP": (45, 250),
    "CON_APP_WEB": (5, 50)
    }

    visitas_web = []
    visitas_app = []

    for tiene_app in tiene_app_descargada:
        
        if tiene_app == 0:
            min_web, max_web = rangos_visitas["NO_APP_WEB"]
            
            mu = (min_web + max_web) / 2
            sigma = (max_web - min_web) / 6
            
            web = int(round(np.clip(random.gauss(mu, sigma), min_web, max_web)))
            
            visitas_web.append(web)
            visitas_app.append(0)
            
        else:
            min_app, max_app = rangos_visitas["CON_APP_APP"]
            mu_app = (min_app + max_app) / 2
            sigma_app = (max_app - min_app) / 6
            app = int(round(np.clip(random.gauss(mu_app, sigma_app), min_app, max_app)))
            visitas_app.append(app)

            min_web, max_web = rangos_visitas["CON_APP_WEB"]
            mu_web = (min_web + max_web) / 2
            sigma_web = (max_web - min_web) / 6
            web = int(round(np.clip(random.gauss(mu_web, sigma_web), min_web, max_web)))
            visitas_web.append(web)

    df_marketing_retail = pd.DataFrame({
    'user_id': ids_clientes,
    'genero': lista_genero,
    'edad': lista_edad,
    'continente': lista_continentes_output,
    'pais': lista_paises_output,
    'fecha_registro': lista_fechas_output,
    'tipo_suscripcion': suscripciones,
    'puntos_cliente': puntos_cliente,
    'medio_pago_guardado': medio_pago_guardado,
    'tiene_tarjeta_retail': tiene_tarjeta_retail,
    'tiene_app_descargada': tiene_app_descargada,
    'fecha_instalacion_app': fecha_instalacion_app,
    'notificaciones_activadas': notificaciones_activadas,
    'plataforma_app': plataforma_app,
    'cantidad_ordenes': cantidad_ordenes,
    'total_gastado': total_gastado,
    'ticket_promedio': ticket_promedio,
    'visitas_web': visitas_web,
    'visitas_app': visitas_app,
    'nivel_educativo': nivel_educativo
    })

    df_marketing_retail.to_excel("mkt_retail_data.xlsx", index=False, sheet_name='mkt_data')
    df_marketing_retail.to_csv("mkt_retail_data.csv", index=False)
    end = time.time()
    print(f" - Tiempo de ejecución  : {end - start:.2f} segundos")
    print(f" - Cantidad de registros: {df_marketing_retail.shape[0]}")
    print(f" - Cantidad de features : {df_marketing_retail.shape[1]}")
    print("------------------------------")
    print("TIPO DE DATO:")
    print(df_marketing_retail.dtypes)
    return df_marketing_retail


mkt_data_generator(num_clientes= 35000, semilla= None)
```
