```python
# Cargamos los archivos
patients = pd.read_csv("patients_dim.csv")
diagnoses = pd.read_csv("diagnoses_dim.csv")
appointments = pd.read_csv("emergency_appointments.csv")

def informe_calidad_detallado(df, nombre_tabla):
    print(f"=== ANÁLISIS DE COLUMNAS CON FALTANTES: {nombre_tabla} ===")
    
    # 1. Identificar Nulos reales (NaN)
    nulos = df.isnull().sum()
    
    # 2. Identificar "Vacíos Fantasma" (espacios en blanco, strings vacíos)
    # Convertimos a string, quitamos espacios y comparamos con vacío
    vacios = df.apply(lambda x: x.astype(str).str.strip().eq('').sum())
    
    # 3. Identificar valores 'None' o 'null' escritos como texto (común en CSVs mal exportados)
    texto_null = df.apply(lambda x: x.astype(str).str.lower().isin(['none', 'null', 'nan']).sum())
    
    # Combinamos todo
    resumen = pd.DataFrame({
        'Nulos_NaN': nulos,
        'Vacios_Texto': vacios,
        'Palabra_Null': texto_null
    })
    
    # Creamos un total
    resumen['Total_Faltantes'] = resumen.sum(axis=1)
    
    # FILTRO CRÍTICO: Solo mostrar las columnas que tienen al menos 1 error
    columnas_con_problemas = resumen[resumen['Total_Faltantes'] > 0]
    
    if columnas_con_problemas.empty:
        print("✅ ¡Felicidades! No se encontraron campos vacíos en esta tabla.\n")
    else:
        print(columnas_con_problemas)
        print("-" * 50)
        # Te dice cuáles son las columnas específicas para que las revises
        print(f"Acción recomendada: Revisar {list(columnas_con_problemas.index)}\n")

# Ejecutamos para tus 3 tablas
listado_dfs = [(patients, "Patients"), (diagnoses, "Diagnoses"), (appointments, "Appointments")]

for df, nombre in listado_dfs:
    informe_calidad_detallado(df, nombre)
```
